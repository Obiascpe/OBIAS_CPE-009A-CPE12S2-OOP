### **PYTHON BUILT IN DATA TYPES**

```python
pi = 3.14159
varlist = [1, 2, 'A', 'B', "Hello!", pi]
print(varlist[0])
```

```python
print(varlist[4])
```

```python
varlist.append("World!")
print(varlist[6])
```

```python
len(varlist)
```

```python
print(varlist[5])
```

```python
varlist.remove(pi)
print(varlist[5])
```

### **DICTIONARY**

```python
var = "Hello World!"
varDict = {'first':123, 2:'abc', '3':var, 4:['Lista', 'Listb']}
print(varDict[2])
```

```python
print(varDict[2])
```

```python
print(varDict['3'])
```

```python
print(varDict[4])
```

```python
print(varDict[4][1])
```

```python
len(varDict)
```

### **PYTHON BUILT IN DATA TYPES**
