Write a function that checks if a string of parentheses (), brackets [], and braces {} is balanced.

Examples:

    "({[]})" → True

    "([)]" → False

    "((()))" → True

    "({)}" → False


```python
#Question 24
def check_ifbalanced(symbol):
    stack = []
    for check in symbol:
        if check in "([{":
            stack += [check]
        else:
            if not stack:
                return False
            top = stack[-1]
            if (check == ")" and top == "(") or \
               (check == "]" and top == "[") or \
               (check == "}" and top == "{"):
                stack = stack[:-1]
            else:
                return False
    if stack:
        return False
    return True

checking = input("Input a string of parenthesis/brackets/braces: ")
print(check_ifbalanced(checking))
```
