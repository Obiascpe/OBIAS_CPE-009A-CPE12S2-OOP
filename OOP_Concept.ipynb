```
#Car's Class Name and Mileage
class Car:
   def __init__(self,name,mileage):
    self.name = name
    self.mileage = mileage
obj1 = Car("Toyota",2202)
obj2 = Car("Subaru",1200)
print(obj1.name)
print(obj1.mileage, "km/l")
print(obj2.name)
print(obj2.mileage, "km/l")
```

```
#Car's Class Name and Mileage with Description
class Car:
   def __init__(self,name,mileage):
    self.name = name
    self.mileage = mileage
obj1 = Car("Toyota",2202)
obj2 = Car("Subaru",1200)
print(obj1.name,obj1.mileage)
print(obj2.name,obj2.mileage)
def description(self):
    return f"The {self.name} car gives the mileage of {self.mileage}km/l"
print(description(obj1))
print(description(obj2))
```

```
#Car's Class Name and Max Speed with Description
class Car:
 def __init__(self,name,speed):
    self.name = name
    self.speed = speed
obj1 = Car("Bugatti",320)
obj2 = Car("Subaru",210)
print(obj1.name,obj1.speed)
print(obj2.name,obj2.speed)
def max_speed(self,speed):
    return f"The {self.name} runs at the maximum speed of {speed}km/hr"
print(max_speed(obj1,320))
print(max_speed(obj2,210))
```

```
#Animal's Features like name, habitat, and temperature
class Animals:
  def __init__(self,name,habitat,temperature):
    self.name = name
    self.habitat = habitat
    self.temperature = temperature


  def features(self):
    return f"The {self.name} is a {self.temperature}-blooded animal that lives on {self.habitat}."

mammals = Animals("Dog", "Land", "Warm")
birds = Animals("Sparrow", "Air", "Warm")
reptiles = Animals("Snake", "Terrestrial", "Cold")

print(mammals.name)
print(mammals.habitat)

print(birds.name)
print(birds.habitat)

print(reptiles.name)
print(reptiles.habitat)

print(mammals.features())
print(birds.features())
print(reptiles.features())
```

```
import math

class Shape:

    def __init__(self, name, sides):
        self.name = name
        self.sides = sides

    def description(self):
        return "The {} has {} sides.".format(self.name, self.sides)

    def area_circle(self, radius):
        return 3.14 * radius * radius

    def area_square(self, side):
        return side * side

    def area_rectangle(self, length, width):
        return length * width

    def area_triangle(self, base, height):
        return 0.5 * base * height

obj1 = Shape("Circle", 0)
obj2 = Shape("Square", 4)
obj3 = Shape("Rectangle", 4)
obj4 = Shape("Triangle", 3)

print(obj1.description())
print("Area of {}: {}".format(obj1.name, obj1.area_circle(5)))

print(obj2.description())
print("Area of {}: {}".format(obj2.name, obj2.area_square(10)))

print(obj3.description())
print("Area of {}: {}".format(obj3.name, obj3.area_rectangle(5, 10)))

print(obj4.description())
print("Area of {}: {}".format(obj4.name, obj4.area_triangle(10, 8)))
```
